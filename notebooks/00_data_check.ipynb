```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/processed/matches.csv', parse_dates=['Date'])
print('shape:', df.shape)
print('date range:', df['Date'].min().date(), 'to', df['Date'].max().date())
df.head()
```

```python
g = df.assign(goals=df.FTHG + df.FTAG).groupby('Season')['goals'].mean()
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(g.index, g.values, marker='o')
ax.set_xlabel('Season'); ax.set_ylabel('Mean goals per match')
ax.set_title('Mean total goals per match by season (2026/27 partial)')
ax.tick_params(axis='x', rotation=45); ax.grid(alpha=.3)
plt.tight_layout(); plt.show()
```
